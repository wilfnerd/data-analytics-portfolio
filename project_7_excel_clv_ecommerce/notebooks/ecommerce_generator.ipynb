```python
import pandas as pd
import numpy as np

# Setting seed for reproducibility
np.random.seed(42)
n_rows = 10000

# Generating core fields
customer_ids = [f"CUST-{i}" for i in np.random.randint(1000, 9999, n_rows)]
order_ids = [f"ORD-{i}" for i in range(n_rows)]
order_dates = pd.to_datetime("2022-01-01") + pd.to_timedelta(
    np.random.randint(0, 730, n_rows), unit="D"
)

# Categories and Segments
regions = np.random.choice(["North America", "Europe", "Asia", "Africa"], n_rows)
segments = np.random.choice(["Low Value", "Mid Value", "High Value"], n_rows)

# Financial logic
quantity = np.random.randint(1, 6, n_rows)
price = np.random.uniform(20, 500, n_rows)
revenue = quantity * price
cogs = revenue * np.random.uniform(0.45, 0.7, n_rows)
marketing_cost = np.random.uniform(5, 80, n_rows)
profit = revenue - cogs - marketing_cost

# Assemble the DataFrame
df = pd.DataFrame({
    "Customer_ID": customer_ids,
    "Order_ID": order_ids,
    "Order_Date": order_dates,
    "Region": regions,
    "Customer_Segment": segments,
    "Quantity": quantity,
    "Price_per_Unit": price.round(2),
    "Revenue": revenue.round(2),
    "COGS": cogs.round(2),
    "Marketing_Cost": marketing_cost.round(2),
    "Profit": profit.round(2)
})

# SAVE TO DATA FOLDER
df.to_csv("../data/ecommerce_transactions.csv", index=False)

print("✅ Success! Dataset created at: ../data/ecommerce_transactions.csv")

# DISPLAY FIRST 5 ROWS
df.head()
```
